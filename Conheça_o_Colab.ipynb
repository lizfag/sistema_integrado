```
def cadastro_alunos():
    alunos = []

    while True:
        print("\n" + "="*50)
        print("SISTEMA DE CADASTRO DE ALUNOS")
        print("="*50)
        print("1. Cadastrar aluno")
        print("2. Listar alunos")
        print("3. Buscar aluno pelo nome")
        print("4. Remover aluno")
        print("5. Verificar se nome existe")
        print("6. Comparar nomes (maior e menor)")
        print("7. Transformar nomes em maiúsculas")
        print("8. Remover último aluno")
        print("9. Voltar ao menu principal")

        opcao = input("\nEscolha uma opção: ")

        if opcao == "1":
            nome = input("Digite o nome do aluno: ")
            alunos.append(nome)
            print(f"Aluno '{nome}' cadastrado com sucesso!")

        elif opcao == "2":
            if not alunos:
                print("Nenhum aluno cadastrado.")
            else:
                print("\nLISTA DE ALUNOS:")
                for i, aluno in enumerate(alunos, 1):
                    print(f"{i}. {aluno}")

        elif opcao == "3":
            nome_busca = input("Digite o nome para buscar: ")
            encontrados = [aluno for aluno in alunos if nome_busca.lower() in aluno.lower()]
            if encontrados:
                print("\nALUNOS ENCONTRADOS:")
                for aluno in encontrados:
                    print(f"- {aluno}")
            else:
                print("Nenhum aluno encontrado com esse nome.")

        elif opcao == "4":
            nome_remover = input("Digite o nome do aluno a remover: ")
            if nome_remover in alunos:
                alunos.remove(nome_remover)
                print(f"Aluno '{nome_remover}' removido com sucesso!")
            else:
                print("Aluno não encontrado.")

        elif opcao == "5":
            nome_verificar = input("Digite o nome para verificar: ")
            if nome_verificar in alunos:
                print(f"O nome '{nome_verificar}' existe na lista.")
            else:
                print(f"O nome '{nome_verificar}' NÃO existe na lista.")

        elif opcao == "6":
            if alunos:
                maior_nome = max(alunos, key=len)
                menor_nome = min(alunos, key=len)
                print(f"Maior nome: '{maior_nome}' ({len(maior_nome)} caracteres)")
                print(f"Menor nome: '{menor_nome}' ({len(menor_nome)} caracteres)")
            else:
                print("Nenhum aluno cadastrado.")

        elif opcao == "7":
            if alunos:
                print("\nNOMES EM MAIÚSCULAS:")
                for aluno in alunos:
                    print(f"- {aluno.upper()}")
            else:
                print("Nenhum aluno cadastrado.")

        elif opcao == "8":
            if alunos:
                removido = alunos.pop()
                print(f"Último aluno removido: '{removido}'")
            else:
                print("Nenhum aluno para remover.")

        elif opcao == "9":
            break
        else:
            print("Opção inválida!")

def controle_despesas():
    despesas = []

    print("\n" + "="*50)
    print("CONTROLE DE DESPESAS PESSOAIS")
    print("="*50)
    print("Digite os valores das despesas (0 para encerrar)")

    while True:
        try:
            valor = float(input("Valor da despesa: R$ "))
            if valor == 0:
                break
            if valor < 0:
                print("Digite apenas valores positivos!")
                continue
            despesas.append(valor)
        except ValueError:
            print("Digite um valor numérico válido!")

    if despesas:
        total = sum(despesas)
        media = total / len(despesas)
        maior = max(despesas)
        menor = min(despesas)

        print("\n" + "="*50)
        print("RELATÓRIO DE DESPESAS")
        print("="*50)
        print(f"Total de despesas: R$ {total:.2f}")
        print(f"Média de gastos: R$ {media:.2f}")
        print(f"Maior valor: R$ {maior:.2f}")
        print(f"Menor valor: R$ {menor:.2f}")
        print(f"Quantidade de despesas: {len(despesas)}")
    else:
        print("Nenhuma despesa registrada.")

def cardapio_lanchonete():
    itens = ["Hambúrguer", "Cheeseburger", "Batata Frita", "Refrigerante", "Suco", "Sorvete"]
    precos = [15.90, 17.90, 12.50, 8.00, 10.00, 9.50]
    pedido = []

    while True:
        print("\n" + "="*50)
        print("CARDÁPIO DA LANCHONETE")
        print("="*50)
        for i, (item, preco) in enumerate(zip(itens, precos), 1):
            print(f"{i}. {item:.<30} R$ {preco:.2f}")
        print("0. Finalizar pedido")

        try:
            escolha = int(input("\nEscolha o item (número): "))
            if escolha == 0:
                break
            if 1 <= escolha <= len(itens):
                pedido.append((itens[escolha-1], precos[escolha-1]))
                print(f"'{itens[escolha-1]}' adicionado ao pedido!")
            else:
                print("Opção inválida!")
        except ValueError:
            print("Digite um número válido!")

    if pedido:
        print("\n" + "="*50)
        print("RESUMO DO PEDIDO")
        print("="*50)
        total = 0
        for item, preco in pedido:
            print(f"{item:.<30} R$ {preco:.2f}")
            total += preco
        print("-"*50)
        print(f"TOTAL: {'':.<23} R$ {total:.2f}")
    else:
        print("Nenhum item foi pedido.")

def controle_estoque():
    produtos = []
    quantidades = []

    while True:
        print("\n" + "="*50)
        print("CONTROLE DE ESTOQUE - LANCHONETE")
        print("="*50)
        print("1. Adicionar produto")
        print("2. Atualizar quantidade")
        print("3. Listar todos os produtos")
        print("4. Mostrar produtos com estoque zero")
        print("5. Voltar ao menu principal")

        opcao = input("\nEscolha uma opção: ")

        if opcao == "1":
            produto = input("Nome do produto: ")
            try:
                quantidade = int(input("Quantidade: "))
                produtos.append(produto)
                quantidades.append(quantidade)
                print(f"Produto '{produto}' adicionado com sucesso!")
            except ValueError:
                print("Digite uma quantidade válida!")

        elif opcao == "2":
            if not produtos:
                print("Nenhum produto cadastrado.")
                continue

            print("\nPRODUTOS CADASTRADOS:")
            for i, (prod, qtd) in enumerate(zip(produtos, quantidades), 1):
                print(f"{i}. {prod} - Estoque: {qtd}")

            try:
                num = int(input("\nNúmero do produto para atualizar: "))
                if 1 <= num <= len(produtos):
                    nova_qtd = int(input("Nova quantidade: "))
                    quantidades[num-1] = nova_qtd
                    print(f"Estoque de '{produtos[num-1]}' atualizado para {nova_qtd}")
                else:
                    print("Número inválido!")
            except ValueError:
                print("Digite um número válido!")

        elif opcao == "3":
            if not produtos:
                print("Nenhum produto cadastrado.")
            else:
                print("\nESTOQUE COMPLETO:")
                for produto, quantidade in zip(produtos, quantidades):
                    print(f"- {produto}: {quantidade} unidades")

        elif opcao == "4":
            zerados = [(prod, qtd) for prod, qtd in zip(produtos, quantidades) if qtd == 0]
            if zerados:
                print("\nPRODUTOS COM ESTOQUE ZERADO:")
                for produto, quantidade in zerados:
                    print(f"- {produto}")
            else:
                print("Nenhum produto com estoque zerado.")

        elif opcao == "5":
            break
        else:
            print("Opção inválida!")

def main():
    while True:
        print("\n" + "="*60)
        print("SISTEMA INTEGRADO - MENU PRINCIPAL")
        print("="*60)
        print("1. Sistema de Cadastro de Alunos")
        print("2. Controle de Despesas Pessoais")
        print("3. Cardápio da Lanchonete")
        print("4. Controle de Estoque da Lanchonete")
        print("5. Sair")

        opcao = input("\nEscolha uma opção: ")

        if opcao == "1":
            cadastro_alunos()
        elif opcao == "2":
            controle_despesas()
        elif opcao == "3":
            cardapio_lanchonete()
        elif opcao == "4":
            controle_estoque()
        elif opcao == "5":
            print("Saindo do sistema... Até logo!")
            break
        else:
            print("Opção inválida! Tente novamente.")

if __name__ == "__main__":
    main()
```
